```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_csv(r'C:\Users\user\Documents\Datascience\group_13_projects\Medical-Insurance-Cost-Prediction-System\data\insurance_clean.csv')

# Quick overview of the dataset
print(f"First few rows of the dataset:\n{df.head()}")
print(f"\nDataset information:\n{df.info()}")
```

```python
print(df.describe())               # summary stats for numeric columns
print(df['smoker'].value_counts()) # how many smokers vs non smokers
print(df['sex'].value_counts())    # male vs female count
print(df['region'].value_counts()) # region breakdown
```

```python
#Analyze the distributions

# Distribution of Charges (target variable)
plt.figure(figsize=(8,5))
sns.histplot(df['charges'], kde=True, color='steelblue')
plt.title('Distribution of Insurance Charges')
plt.xlabel('Charges ($)')
plt.show()
```

```python
# Distribution of Age
plt.figure(figsize=(8,5))
sns.histplot(df['age'], kde=True, color='green')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.show()
```

```python
# Distribution of BMI
plt.figure(figsize=(8,5))
sns.histplot(df['bmi'], kde=True, color='orange')
plt.title('Distribution of BMI')
plt.xlabel('BMI')
plt.show()
```

```python
# Categorical Distributions
print("\nSmoker Count:")
print(df['smoker'].value_counts())

print("\nSex Count:")
print(df['sex'].value_counts())

print("\nRegion Count:")
print(df['region'].value_counts())

print("\nChildren Count:")
print(df['children'].value_counts())
```

```python
# EXPLORE CORRELATIONS

# Heatmap - numeric columns only
plt.figure(figsize=(8,6))
sns.heatmap(df[['age','bmi','children','charges']].corr(), 
            annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix - Numeric Variables')
plt.show()
```

```python
#IDENTIFY PATTERNS AND TRENDS
# Smoker vs Charges (most impactful variable)
plt.figure(figsize=(8,5))
sns.boxplot(x='smoker', y='charges', data=df, palette='Set2')
plt.title('Smoker vs Insurance Charges')
plt.xlabel('Smoker')
plt.ylabel('Charges ($)')
plt.show()
```

```python
# Age vs Charges
plt.figure(figsize=(8,5))
sns.scatterplot(x='age', y='charges', hue='smoker', data=df, palette='Set1')
plt.title('Age vs Insurance Charges by Smoking Status')
plt.xlabel('Age')
plt.ylabel('Charges ($)')
plt.show()
```

```python
# BMI vs Charges
plt.figure(figsize=(8,5))
sns.scatterplot(x='bmi', y='charges', hue='smoker', data=df, palette='Set1')
plt.title('BMI vs Insurance Charges by Smoking Status')
plt.xlabel('BMI')
plt.ylabel('Charges ($)')
plt.show()
```

```python
# Region vs Charges
plt.figure(figsize=(8,5))
sns.boxplot(x='region', y='charges', data=df, palette='Set3')
plt.title('Region vs Insurance Charges')
plt.xlabel('Region')
plt.ylabel('Charges ($)')
plt.show()
```

```python
# Sex vs Charges
plt.figure(figsize=(8,5))
sns.boxplot(x='sex', y='charges', data=df, palette='Set2')
plt.title('Sex vs Insurance Charges')
plt.xlabel('Sex')
plt.ylabel('Charges ($)')
plt.show()
```

```python
# Children vs Charges
plt.figure(figsize=(8,5))
sns.boxplot(x='children', y='charges', data=df, palette='Set3')
plt.title('Number of Children vs Insurance Charges')
plt.xlabel('Number of Children')
plt.ylabel('Charges ($)')
plt.show()
```

```python
# 5. SUMMARY / FINDINGS

print("\nCorrelation of all numeric variables with Charges:")
print(df[['age','bmi','children','charges']].corr()['charges'].sort_values(ascending=False))

print("\nAverage Charges by Smoker:")
print(df.groupby('smoker')['charges'].mean())

print("\nAverage Charges by Region:")
print(df.groupby('region')['charges'].mean())

print("\nAverage Charges by Sex:")
print(df.groupby('sex')['charges'].mean())
```

```python
#FINDINGS:
# 1. Smoker status is the strongest predictor - smokers pay 
#    almost 4x more than non-smokers ($32,050 vs $8,440)
# 2. Age has a moderate positive correlation with charges (0.30)
# 3. BMI has a weak positive correlation with charges (0.20)
# 4. Region and sex have minimal impact on charges
# 5. Number of children has almost no impact (0.07)
```
